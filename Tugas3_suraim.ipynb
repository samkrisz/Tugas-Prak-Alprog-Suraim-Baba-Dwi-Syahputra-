1.) Membuat Data Belanjaan
Doraemon ingin mulai dengan mencatat barang belanjaan Nobita. <br>

• Buat sebuah list belanjaan yang berisi tiga barang: **“Apel”**, **“Susu”**,
dan **“Roti”**. <br>

• Buat sebuah tuple harga_jumlah yang berisi harga dan jumlah
masing-masing barang dengan format:

image.png

```python

belanjaan = ["Apel", "Susu", "Roti"]          # Membuat list berisi tiga barang
harga_jumlah = (5000, 3, 12000, 2, 8000, 1)

print("Daftar belanjaan:", belanjaan)
print("Harga dan jumlah:", harga_jumlah)

print("Karakter pertama:", belanjaan[0][0])
print("Karakter terakhir:", belanjaan[-1][-1])
```

2.) **Hitung Total Harga per Barang** <br>
Setelah daftar belanjaan dicatat, sekarang saatnya menghitung harga. <br>

• Hitung total harga masing-masing barang secara manual.   
• Hitung total keseluruhan belanja dengan menjumlahkan total per
barang.

```python

harga_apel = harga_jumlah[0]
jumlah_apel = harga_jumlah[1]

harga_susu = harga_jumlah[2]
jumlah_susu = harga_jumlah[3]

harga_roti = harga_jumlah[4]
jumlah_roti = harga_jumlah[5]

total_apel = harga_apel * jumlah_apel
total_susu = harga_susu * jumlah_susu
total_roti = harga_roti * jumlah_roti

print(f"Total harga Apel: {total_apel}")
print(f"Total harga Susu: {total_susu}")
print(f"Total harga Roti: {total_roti}")

total_belanja = total_apel + total_susu + total_roti

print(f"Total seluruh belanja: {total_belanja}")
```

3.) **Cetak Struk Belanja**<br>
Setelah semua data belanjaan dicatat, Doraemon ingin menyimpannya
dalam variabel agar lebih mudah dipahami. Data belanjaan Nobita
sebagai berikut: <br>

• Nama pembeli = “Nobita Nobi” <br>
• Nama toko = “Toko Sembako Doraemon” <br>
• List **belanjaan** dan **Tuple harga_jumlah**
Agar Nobita tidak bingung, Doraemon ingin menampilkan struk belanja
yang rapi.<br>
Gunakan f-string dan escape sequences untuk mencetak struk
belanja dengan format sebagai berikut:

image.png

```python

nama_pembeli = "Nobita Nobi"
nama_toko = "Toko Sembako Doraemon"

print("=" * 40)
print(f"{nama_toko:^40}")
print("=" * 40)
print(f"Nama Pembeli : {nama_pembeli}")
print("-" * 40)
print(f"{'Barang':<10}{'Harga':>10}{'\tJumlah':>8}{'Total\t\t':>12}")
print("-" * 40)
print(f"{belanjaan[0]:<10}{harga_jumlah[0]:>10,}{harga_jumlah[1]:>8}{total_apel:>12,}")
print(f"{belanjaan[1]:<10}{harga_jumlah[2]:>10,}{harga_jumlah[3]:>8}{total_susu:>12,}")
print(f"{belanjaan[2]:<10}{harga_jumlah[4]:>10,}{harga_jumlah[5]:>8}{total_roti:>12,}")
print("-" * 40)
print(f"{'TOTAL':<28}Rp{total_belanja:>10,}")
print("=" * 40)
print("     Terima kasih telah belanja!🙏🏻\t")
print("=" * 40)
```

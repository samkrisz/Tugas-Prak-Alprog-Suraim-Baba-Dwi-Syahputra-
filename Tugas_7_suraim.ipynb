a. **Buatlah folder project yang bernama saving_project**. <br>Folder ini akan

b. **Membuat file modul interest.py**<br>
Modul ini hanya berisi fungsi compound_interest dan simple_interest.
Tidak boleh ada input() atau print() pada modul ini tujuannya agar
modul bisa dipakai ulang oleh program lain. <br>

• compoud_interest(): fungsi untuk bunga majemuk<br>
image.png

• simple_interest(): fungsi untuk bunga sederhana <br>

image.png


```python
%%writefile saving_project/interest.py

def compound_interest(principal, rate, time):
    future_value = principal * (1 + rate) ** time
    return future_value


def simple_interest(principal, rate, time):
    future_value = principal + (principal * rate * time)
    return future_value
```

c. **Membuat file modul display.py** <br>
Modul ini hanya berisi fungsi print_result(principal, rate, years, ci, dan
si). Fungsi ini akan menampilkan display Saldo Awal, Bunga Tahunan,
15
Lama Menabung, Nilai Akhir (Bunga Majemuk) dan Nilai Akhir (Bunga
Sederhana).

```python
%%writefile saving_project/display.py

def print_result(principal, rate, years, compound, simple):

    print("\n===== HASIL PERHITUNGAN TABUNGAN =====")

    print(f"Saldo Awal          : ¥{principal:,.1f}")
    print(f"Bunga Tahunan       : {rate * 100:.1f}%")
    print(f"Lama Menabung       : {years} tahun")

    print("------------------------------------------")

    print(f"Nilai Akhir (Bunga Majemuk)   : ¥{compound:,.2f}")
    print(f"Nilai Akhir (Bunga Sederhana) : ¥{simple:,.2f}")

    print("==========================================")
```

d. **Membuat file modul main.py**<br>
Modul ini berisi main program dengan judul “Program Kalkulator
Bunga Tabungan” dan menyediakan input principal, rate dan years.
Pada modul ini akan melakukan import modul interest dan display.

```python
%%writefile saving_project/main.py

from interest import compound_interest, simple_interest
from display import print_result


def main():

    print("=== Program Kalkulator Bunga Tabungan ===")

    # Data sesuai soal Nobita
    principal = 80000
    rate = 0.05
    years = 4

    # Menghitung bunga majemuk
    compound = compound_interest(principal, rate, years)

    # Menghitung bunga sederhana
    simple = simple_interest(principal, rate, years)

    # Menampilkan hasil perhitungan
    print_result(principal, rate, years, compound, simple)


if __name__ == "__main__":
    main()
```

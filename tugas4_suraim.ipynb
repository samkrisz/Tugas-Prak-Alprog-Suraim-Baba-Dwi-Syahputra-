2.)Suatu hari, Nobita baru saja mendapatkan motor listrik hasil ciptaan
Doraemon yang dinamai “DoraBike”. Namun, karena Nobita terkenal
ceroboh dan sering menabrak sesuatu, Doraemon khawatir jika DoraBike
rusak dan biayanya mahal. Akhirnya Doraemon membuat sistem pintar
bernama “Smart Vehicle Insurance Analyzer”, yang bisa menghitung risiko
dan premi asuransi kendaraan berdasarkan kondisi penggunaan motor
listrik Nobita. Program ini akan menilai tingkat risiko asuransi kendaraan
dan premi yang harus dibayar Nobita berdasarkan beberapa parameter
berikut. <br>

```python
print("=== Masukkan Data Penggunaan Kendaraan ===")

usia_pengemudi = int(input("Masukkan usia pengemudi (tahun): "))
jenis_kendaraan = input("Masukkan jenis kendaraan ('motor' atau 'mobil'): ").strip().lower()
nilai_kendaraan = float(input("Masukkan nilai kendaraan (dalam Rupiah): "))
area_penggunaan = input("Masukkan area penggunaan ('perkotaan' atau 'luar perkotaan'): ").strip().lower()
sistem_pengaman = input("Apakah kendaraan memiliki sistem pengaman? ('ya' atau 'tidak'): ").strip().lower()

if usia_pengemudi < 20 or usia_pengemudi > 60:
    risiko = "tinggi"
else:

    if jenis_kendaraan == "motor":
        risiko = "sedang"
    elif jenis_kendaraan == "mobil":
        risiko = "rendah"
    else:
        risiko = "sedang"


if area_penggunaan == "perkotaan":
    if risiko == "rendah":
        risiko = "sedang"
    elif risiko == "sedang":
        risiko = "tinggi"



if sistem_pengaman == "tidak":
    if risiko == "rendah":
        risiko = "sedang"
    elif risiko == "sedang":
        risiko = "tinggi"
    elif risiko == "tinggi":
        risiko = "tinggi"


if risiko == "tinggi":
    persentase_premi = 0.1
elif risiko == "sedang":
    persentase_premi = 0.07
else:
    persentase_premi = 0.05

premi_asuransi = persentase_premi * nilai_kendaraan


pesan = (
    "Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!"
    if risiko == "tinggi"
    else "Risiko kendaraan masih aman menurut Doraemon."
)


print("\n" + "="*40)
print("HASIL ANALISIS SMART VEHICLE INSURANCE")
print("="*40)
print(f"1. Kategori Risiko         : {risiko.capitalize()}")
print(f"2. Premi Asuransi          : Rp {premi_asuransi:,.2f}")
print(f"3. Pesan Doraemon          : {pesan}")
print("="*40)
```

**a.) Mencatat pengeluaran harian Nobita.**<br>
Doraemon ingin mencatat pengeluaran harian Nobita selama
sebulan, tapi pencatatan harus berhenti jika saldo habis. Buatlah
program untuk miminta inputan pengeluaran harian Nobita. Kemudian
nilai saldo akan berkurang setiap kali input dimasukkan & berhenti
(break) jika saldo Nobita < 0. Tampilkan sisa saldo Nobita setiap hari. <br>
• Gunakan variabel hari = 1 dan saldo = 50000. <br>
• Gunakan kondisi while hari ≤ 30: <br>
• Didalam perulangan while:
1. Minta input pengeluaran.
2. Kurangi saldo
3. Gunakan of saldo < 0: untuk menghentikan loop dengan
break.<br>
4. Jika masih cukup, tampilkan sisa saldo dan tambahkan hari
+= 1. <br>

• Tambahkan else: setelah loop untuk mencetak pesan jika
terjadi break.

image.png

```python

hari = 1
saldo = 50000

while hari <= 30:
    pengeluaran = int(input(f"Masukkan pengeluaran hari ke-{hari}: Rp"))

    saldo -= pengeluaran

    if saldo < 0:
        print(f"Saldo habis di hari ke-{hari}. Doraemon menghentikan pencatatan!")
        break

    print(f"Sisa saldo hari ke-{hari}: Rp{saldo}")

    hari += 1

else:
    print("Pencatatan 30 hari selesai.")
```

**b.) Menganalisis pengeluaran mingguan Nobita.**<br>

Doraemon ingin menganalisis pengeluaran Nobita per minggu
selama 4 minggu. Namun, jika pada minggu tertentu Nobita tidak jajan
sama sekali, minggu itu dilewati (tidak dianalisis).<br>
• Buat list: Pengeluaran_mingguan = [20000, 0, 15000, 0]<br>
• Buat perulanagn dengan for i in range(4): <br>
• Gunakan if jika pengeluaran_mingguan == 0, maka continue<br>
• Jika tidak nol, tampilkan pengeluaran minggu itu <br>
• Tambahkan else: setelah perulangan untuk menampilkan
bahwa analisis selesai
image.png

```python

Pengeluaran_mingguan = [20000, 0, 15000, 0]

for i in range(4):
    if Pengeluaran_mingguan[i] == 0:
        print(f"Minggu ke-{i + 1}: Nobita tidak jajan, dilewati analisisnya.")
        continue

    print(f"Minggu ke-{i + 1}: Total pengeluaran Rp{Pengeluaran_mingguan[i]}")

else:
    print("Analisis mingguan selesai.")
```

**c.) Menyusun laporan bulanan Nobita.**<br>
Doraemon ingin membuat simulasi keuangan Nobita selama **3 bulan**.
Setiap bulan terdiri dari **4 minggu**. Jika di salah satu minggu
pengeluaran Nobita > Rp20.000, pencatatan bulan itu dihentikan
(break). <br>
• Gunakan for untuk perulangan outer bulan (1-3) <br>
• Gunakan for untuk perulangan inner minggu (1-4) <br>
• Input pengeluaran tiap minggu <br>
• Tambahkan total bulanan <br>
• Gunakan kondisi jika pengeluaran > 20000 hentikan<br>
• Tambahkan else: setelah perulangan inner untuk mencetak total
pengeluaran bulan itu. <br>
• Setelah semua bulan selesai, tampilkan ringkasan.<br>
image.png

```python

for bulan in range(1, 4):
    print(f"\nBulan ke-{bulan}")

    total_bulanan = 0

    for minggu in range(1, 5):
        pengeluaran = int(
            input(f"Pengeluaran minggu ke-{minggu}: Rp")
        )

        if pengeluaran > 20000:
            print("Pengeluaran terlalu besar! Pencatatan bulan ini dihentikan.")
            break

        total_bulanan += pengeluaran

    else:
        print(f"Total pengeluaran bulan ke-{bulan}: Rp{total_bulanan}")
```

**d.) Evaluasi Akhir Tahun.**<br>
Doraemon ingin mengevaluasi keuangan Nobita selama **12 bulan**.
Setiap bulan terdiri dari **4 minggu**. Jika di suatu minggu adalah hari libur, minggu itu dilewati (continue). Jika saldo < 0, simulasi langsung
berhenti (break). Jika sampai akhir tahun saldo tetap positif, tampilkan
ucapan keberhasilan dengan else: <br>
• Inisasi saldo_awal = 300000 <br>
• Gunakan for untuk perulangan bulan (1-12)<br>
• Gunakan while untuk perluangan minggu (≤ 4) <br>
• Buatkan Inputan apakah minggu tersebut libur (y/n)?<br>
• Jika y : lanjutkan ke minggu berikutnya (continue) <br>
• Jika n : minta input pengeluaran dan kurangi saldo. <br>
• Jika saldo < 0 : break <br>
• Gunakan else setelah while untuk mencetak saldo akhir bulan. <br>
• Gunakan else setelah for untuk mengucapkan “Nobita berhasil <br>
menabung {saldo} dari total {saldo_awal}“ <br>
image.png


```python

saldo_awal = 300000
saldo = saldo_awal

for bulan in range(11, 13):
    print(f"\n=== Bulan ke-{bulan} ===")
    minggu = 1

    while minggu <= 4:
        libur = input(f"Apakah minggu ke-{minggu} libur? (y/n): ")

        if libur.lower() == "y":
            print("Minggu ini dilewati (continue).")
            minggu += 1
            continue

        pengeluaran = int(input("Masukkan pengeluaran minggu ini: "))
        saldo -= pengeluaran

        if saldo < 0:
            print("Saldo habis! Simulasi dihentikan.")
            break

        minggu += 1

    else:
        print(f"Saldo tersisa akhir bulan {bulan}: Rp{saldo}")

    if saldo < 0:
        break

else:
    print(f"\n Nobita berhasil menabung Rp{saldo} dari total Rp{saldo_awal}!")
```

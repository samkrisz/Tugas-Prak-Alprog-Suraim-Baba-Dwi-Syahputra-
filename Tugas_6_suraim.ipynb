**a. Membuat Variabel Global** <br>
Buatlah variabel **global skor_nobita = 0** untuk menyimpan skor nobita
selama simulasi berlangsung

```python
# Variabel global untuk menyimpan skor Nobita
skor_nobita = 0
```

**b. Membuat Fungsi untuk Memperbarui Skor**<br>
Buat fungsi update_skor(nilai) yang akan menambah nilai ke
skor_nobita. Gunakan kata kunci global agar fungsi bisa mengubah
variabel global di luar fungsi.

```python
def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai
```

**c. Membuat Fungsi Aktivitas Nobita**
1. Fungsi belajar(jam_belajar)
Gunakan logika if–elif–else untuk menentukan skor berdasarkan
jam belajar dan gunakan return untuk mengembalikan nilai skor.

Screenshot 2026-11-02 163250.png

```python
def belajar(jam_belajar):

    if jam_belajar < 1:
        skor = 50
        print("Nobita hanya membuka buku sebentar")

    elif jam_belajar < 3:
        skor = 25
        print("Nobita belajar secukupnya")

    else:
        skor = 10
        print("Nobita belajar dengan rajin hari ini")

    return skor
```

**2. Fungsi tidur(jam_tidur)** <br>

 Gunakan if–elif–else untuk mengatur efek waktu tidur dan
 gunakan return untuk mengembalikan nilai skor.
 Screenshot 2026-11-02 163341.png

```python
def tidur(jam_tidur):

    if jam_tidur < 1:
        skor = -10
        print("Nobita kurang tidur")

    elif 1 <= jam_tidur <= 2:
        skor = 20
        print("Waktu tidur Nobita seimbang")

    elif jam_tidur > 3:
        skor = -20
        print("Nobita tidur terlalu lama")

    else:
        skor = 0
        print("Waktu tidur belum memiliki ketentuan skor")

    return skor
```

**3. Fungsi main_bersama_doraemon(jumlah_gadget, waktu_main)** <br>

Gunakan if-elif-else untuk mengatur jumlah gadget dan waktu
main dan gunakan return untuk mengembalikan nilai skor.

image.png

```python
def main_bersama_doraemon(jumlah_gadget, waktu_main):

    if waktu_main <= 1 and jumlah_gadget <= 2:
        skor = 15
        print("Nobita bermain secukupnya")

    elif waktu_main <= 2 and jumlah_gadget <= 3:
        skor = 5
        print("Nobita masih bisa mengatur waktu dengan baik")

    else:
        skor = -30
        print("Nobita kebanyakan bermain")

    return skor
```

**4. Fungsi menyontek()** <br>

Fungsi ini tidak membutuhkan parameter, namun tetap memiliki
nilai kembalian (return) skor. Berikan output : “Nobita ketahuan
menyontek” dan berikan skor (-50).

```python
def menyontek():

    print("Nobita ketahuan menyontek")

    return 0
```

**d. Membuat Fungsi Pengendali Aktivitas**<br>

Gunakan alur kendali (control flow) untuk menampilkan menu dan
memanggil fungsi sesuai pilihan. Beri nama fungsi tersebut jalankan
aktivitas (pilihan).

image.png

```python
def jalankan_aktivitas(pilihan):

    if pilihan == 1:

        jam_belajar = float(input("Berapa jam Nobita belajar hari ini? "))
        nilai = belajar(jam_belajar)
        update_skor(nilai)

    elif pilihan == 2:

        jam_tidur = float(input("Berapa jam Nobita tidur siang hari ini? "))
        nilai = tidur(jam_tidur)
        update_skor(nilai)

    elif pilihan == 3:

        jumlah_gadget = int(input("Berapa gadget yang digunakan Nobita? "))
        waktu_main = float(input("Berapa jam Nobita bermain gadget? "))

        nilai = main_bersama_doraemon(jumlah_gadget, waktu_main)
        update_skor(nilai)

    elif pilihan == 4:

        nilai = menyontek()
        update_skor(nilai)

    else:
        print("Pilihan tidak valid, aktivitas hanya ada 4 pilihan")
```

**e. Membuat Fungsi Evaluasi Akhir**

Gunakan if-elif-else dan return untuk mengembalikan hasil evaluasi
perilaku Nobita.  

image.png

```python
def evaluasi_akhir(skor):

    if skor >= 80:
        return "Siswa Teladan"

    elif 50 <= skor < 80:
        return "Cukup Rajin"

    else:
        return "Pemalas"
```

**f.
Membuat Menu Utama Program Peningkat Disiplin Harian**<br>

Buatlah tampilan menu utama program sebagai berikut:

image.png

```python
def menu_utama():

    print("=== PROGRAM PENINGKAT DISIPLIN NOBITA ===")
    print("1. Belajar")
    print("2. Tidur")
    print("3. Main bersama Doraemon")
    print("4. Menyontek saat ujian")

    pilihan = int(input("\nPilih aktivitas (1-4): "))

    jalankan_aktivitas(pilihan)

    print("\n========== HASIL EVALUASI ==========")
    print("Skor Nobita:", skor_nobita)
    print("Evaluasi:", evaluasi_akhir(skor_nobita))


# Memanggil menu utama
menu_utama()
```

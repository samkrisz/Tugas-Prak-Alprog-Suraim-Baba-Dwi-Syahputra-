5.) Bulan pertama, Toko Dorayaki tersebut telah menjual 350 dorayaki
dengan harga Rp. 5.000 per buah. Buatlah variabel jumlah_dorayaki dan
harga_dorayaki, lalu hitung total pendapatan Toko Dorayaki tersebut.

```python
# Menentukan jumlah dorayaki yang terjual
jumlah_dorayaki = 350

# Menentukan harga satu buah dorayaki
harga_dorayaki = 5000

# Menghitung total pendapatan Toko Dorayaki
total_pendapatan = jumlah_dorayaki * harga_dorayaki

# Menampilkan total pendapatan
print("Total pendapatan Toko Dorayaki adalah Rp", total_pendapatan)
```

4.) Selama diet, Giant punya target minum **2000 ml air/hari, 1 botol = 350 ml.**<br>
• Gunakan true division untuk mengetahui hasil asli <br>
• Gunakan floor division untuk botol penuh <br>
• Gunakan modulo untuk sisa air<br>

```python
target_air = 2000
kapasitas_botol  = 350
print("Hasil true division:", target_air / kapasitas_botol)
print("Botol penuh:", target_air // kapasitas_botol)
print("Sisa air:", target_air % kapasitas_botol, "ml")
```
